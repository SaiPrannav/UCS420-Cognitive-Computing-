## Q2. Tuples

**2.** Create a tuple named `scores` containing 8 marks: use the first 8 values from your list L in Q1.

```python
digits = [int(d) for d in "1024170195"]
L_original = [d * 10 for d in digits]

scores = tuple(L_original[:8])
print("scores =", scores)
```

**i.** Find the highest score and its index in the tuple, and the lowest score and how many times it appears.

```python
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)

print(f"Highest score: {highest} at index {highest_index}")
print(f"Lowest score: {lowest}, appears {lowest_count} time(s)")
```

## Q3. Random Numbers

**3.** Set the random seed to your own roll number using `random.seed(your_roll_number)` before generating your numbers.

```python
import random

random.seed(1024170195)
random_list = [random.randint(100, 900) for _ in range(100)]
print(random_list)
```

**i.** Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
print("List already generated above, length:", len(random_list))
```

**ii.** Count and print all odd numbers in the list.

```python
odd_count = sum(1 for x in random_list if x % 2 != 0)
print("Count of odd numbers:", odd_count)
```

**iii.** Count and print all even numbers in the list.

```python
even_count = sum(1 for x in random_list if x % 2 == 0)
print("Count of even numbers:", even_count)
```

**iv.** Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_list if is_prime(x)]
print("Count of prime numbers:", len(prime_numbers))
print("Prime numbers found:", prime_numbers)
```

**v.** Print the number that occurs most frequently in your list, and how many times it occurs.

```python
from collections import Counter

counts = Counter(random_list)
most_common_num, freq = counts.most_common(1)[0]
print(f"Most frequent number: {most_common_num}, occurs {freq} time(s)")
```

## Q4. Sets

**4.** Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

```python
digits_8 = [int(d) for d in "1024170195"][:8]

A = {d * 7 for d in digits_8}
B = {d * 9 for d in digits_8}

print("Digits used:", digits_8)
print("A =", A)
print("B =", B)
```

**vi.** Find the union of A and B — all unique values across both sets.

```python
union_set = A.union(B)
print("A ∪ B =", union_set)
```

**vii.** Find the intersection of A and B — values common to both.

```python
intersection_set = A.intersection(B)
print("A ∩ B =", intersection_set)
```

**viii.** Find A − B and B − A separately (not combined) using `difference()`, and explain in one line how this differs from `symmetric_difference()`.

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B =", A_minus_B)
print("B - A =", B_minus_A)
```

**ix.** Find the symmetric difference of A and B.

```python
sym_diff = A.symmetric_difference(B)
print("A △ B =", sym_diff)
```

**x.** Check whether A is a subset of B, and whether B is a superset of A, using `issubset()` and `issuperset()`.

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

**xi.** Ask the user to input a value X. Remove X from set A if present using `discard()` (not `remove()`) — explain in one line why `discard()` is safer than `remove()` when you're not sure the value exists.

```python
X = int(input("Enter a value to discard from A: "))
A.discard(X)
print("A after discard:", A)
```
